```python
LAST_NAME = "Parras"
STUDENT_ID = "TUPM-26-3579"
FAVORITE_ARTIST = "Daughtry"

SEED_DIGIT = int(STUDENT_ID[-1]) 
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)

# Generate Order Dataset
orders = [
    ID_SUM + SEED_DIGIT,
    ID_SUM + NAME_LENGTH,
    ID_SUM + ARTIST_LENGTH,
    ID_SUM + SEED_DIGIT + NAME_LENGTH,
    ID_SUM + ARTIST_LENGTH + NAME_LENGTH,
    ID_SUM * 2 + SEED_DIGIT,
    ID_SUM * 2 + NAME_LENGTH,
    ID_SUM * 2 + ARTIST_LENGTH
]

print("GENERATED ORDER DATASET:", orders)
```

```python
#orders
sorted_orders = sorted(orders)
reverse_orders = sorted_orders[::-1]
random_orders = orders.copy()

import random
random.seed(SEED_DIGIT)
random.shuffle(random_orders)

print("Sorted Orders:", sorted_orders)
print("Reversed Orders:", reverse_orders)
print("Random Orders:", random_orders)
```

```python
def bubble_sort(data):
    arr = data.copy()
    comparisons = 0
    movements = 0


    for i in range(len(arr)):
        for j in range(len(arr) - 1 - i):
            comparisons += 1

            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                movements += 1

    return arr, comparisons, movements

conditions = {
    "Sorted": sorted_orders,
    "Reverse-Ordered": reverse_orders,
    "Random": random_orders
}

print("BUBBLE SORT")
for name, data in conditions.items():
    result, comp, moves = bubble_sort(data)
    print(f"\n{name} Input:", data)
    print("Sorted Output:", result)
    print("Comparisons:", comp)
    print("Movements:", moves)
    print("Verified:", result == sorted(data))
```

```python
# Selection Sort
def selection_sort(data):
    arr = data.copy()
    comparisons = 0
    movements = 0

    for i in range(len(arr) - 1):
        min_index = i

        for j in range(i + 1, len(arr)):
            comparisons += 1

            if arr[j] < arr[min_index]:
                min_index = j

        if min_index != i:
            arr[i], arr[min_index] = arr[min_index], arr[i]
            movements += 1

    return arr, comparisons, movements


# Print Selection Sort Results
result, comparisons, movements = selection_sort(random_orders)

for name, data in conditions.items():
    result, comp, moves = selection_sort(data)

    print(f"\n{name} Input:", data)
    print("Sorted Output:", result)
    print("Comparisons:", comp)
    print("Movements:", moves)
    

print("SELECTION SORT")

```

```python

# insert sort
def insertion_sort(data):
    arr = data.copy()
    comparisons = 0
    movements = 0

    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0:
            comparisons += 1

            if arr[j] > key:
                arr[j + 1] = arr[j]
                movements += 1
                j -= 1
            else:
                break

        arr[j + 1] = key

    return arr, comparisons, movements


result, comparisons, movements = insertion_sort(random_orders)

print("INSERTION SORT")
print("Original Data:", random_orders)
print("Sorted Output:", result)
print("Comparisons:", comparisons)
print("Movements:", movements)
```

```python
print("SORTING ALGORITHM COMPARISON")
print("Original Data:", random_orders)

for name, sort_function in [
    ("Bubble Sort", bubble_sort),
    ("Selection Sort", selection_sort),
    ("Insertion Sort", insertion_sort)
]:
    result, comparisons, movements = sort_function(random_orders)

    print(f"\n{name}")
    print("Sorted Output:", result)
    print("Comparisons:", comparisons)
    print("Movements:", movements)
    print("Verified:", result == sorted(random_orders))
```
