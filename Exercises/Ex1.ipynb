```python
LAST_NAME = "Parras"
STUDENT_ID = "TUPM-26-3597"
FAVORITE_ARTIST = "Daughtry"

SEED_DIGIT = int(STUDENT_ID[-1]) 
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(ord(char) - ord("A") + 1 for char in FAVORITE_ARTIST.upper())

incident_records = [
    ID_SUM + NAME_LENGTH,
    ID_SUM + ARTIST_LENGTH,
    ID_SUM + ARTIST_SUM,
    ID_SUM + SEED_DIGIT,
    ID_SUM + NAME_LENGTH + ARTIST_LENGTH,
    ID_SUM + ARTIST_SUM + SEED_DIGIT,
    ID_SUM + NAME_LENGTH + ARTIST_SUM,
    ID_SUM + ARTIST_LENGTH + SEED_DIGIT
]


incident_records = sorted(set(incident_records))

print("ORIGINAL INCIDENT RECORDS:")
print(incident_records)
```

```python
# Rotate 
rotation_point = SEED_DIGIT % len(incident_records)

rotated_records = (
    incident_records[rotation_point:]
    + incident_records[:rotation_point]
)

print("ROTATION POINT:", rotation_point)
print("ROTATED INCIDENT RECORDS:", rotated_records)
```

```python

target_incident = ID_SUM + ARTIST_SUM

print("TARGET INCIDENT:", target_incident)
```

```python
# linear search
linear_comparisons = 0
linear_position = -1

for i, incident in enumerate(rotated_records):
    linear_comparisons += 1

    if incident == target_incident:
        linear_position = i
        break

print("LINEAR SEARCH")
if linear_position != -1:
    print("Target found at index:", linear_position)
else:
    print("Target not found.")

print("Comparisons:", linear_comparisons)
```

```python
# binary
binary_comparisons = 0
binary_position = -1

left = 0
right = len(incident_records) - 1

while left <= right:
    mid = (left + right) // 2
    binary_comparisons += 1

    if incident_records[mid] == target_incident:
        binary_position = mid
        break
    elif incident_records[mid] < target_incident:
        left = mid + 1
    else:
        right = mid - 1

print("BINARY SEARCH")
if binary_position != -1:
    print("Target found at index:", binary_position)
else:
    print("Target not found.")

print("Comparisons:", binary_comparisons)
```

```python
# compare each
print("SEARCH COMPARISON")
print("Linear Search Comparisons:", linear_comparisons)
print("Binary Search Comparisons:", binary_comparisons)

print("\nTIME AND SPACE COMPLEXITY")
print("Linear Search: O(n) time, O(1) space")
print("Binary Search: O(log n) time, O(1) space")
```
